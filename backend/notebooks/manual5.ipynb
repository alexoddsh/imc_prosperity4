```python
import cvxpy as cp
import numpy as np
```

```python
products = ['obsidian', 'cells', 'thermalite', 'lava cake', 'magma ink', 'scoria paste', 'ashes', 'volcanic incense', 'sulfur reactor']
man_rets = [0.03, 0.30, 0.40, -0.50, 0.10, 0.0, -0.05, -0.30, 0]
rets = np.array(man_rets)
pi = cp.Variable(9)
objective = cp.Maximize((1_000_000 * rets.T @ pi) - (cp.sum_squares(pi) * 1_000_000))
constraints = [cp.norm(pi, 1) <= 100]
prob = cp.Problem(objective, constraints)

prob.solve()
print('Optimal allocation without integer constraints:')
sum = 0
fees = 0
profit = 0
for i in range(9):
    print("Position in ", products[i], ': ', f"{pi.value[i]:,.2f}", '%', ' | Given expected move: ', man_rets[i], sep='')
    sum += abs(pi.value[i])
    fees += (pi.value[i]**2)*1_000_000
    profit += (pi.value[i])*1_000_000*man_rets[i]

print(f"Total allocation: {sum}")
print(fees)
print(profit-fees)
```
