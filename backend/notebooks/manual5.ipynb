We get maximization problem:

$$max \sum (B \cdot\mu_i w_i - B{w_i^2})$$

Where:
- $\mu_i$ = Expected Return (decimal)
- $w_i$ = Volume allocation (decimal of budget)

Since the expected return is a given exogenus paramater the whole expression depends on one variable. Since the totals are basically guaranteed to not go above 100% since the fees become enormous we can treat this as a unconstrained optimization thus we can simply treat each term separetely. Apply the FOC:

$$\frac{\partial w_i}{\partial F(w_i)} = 0$$

Doing so gives:

$$w_i^* = \mu_i / 2$$

```python
## Predictions
import numpy as np
import pandas as pd

res = pd.DataFrame()

predictions = {
    "PYROFELX": 0.40, 
    "THERMALITE": 0.30,
    "MAGMA_INK": 0.20, 
    "VOLCANIC_INCENSE": 0.10,

    "SULFUR_REACTOR": 0.00,
    "OBSIDIAN_CUTELRY": 0.00,
    "SCORIA_PASTE": 0.00,

    "LAVA_CAKES": -0.50,
    "ASHES_PHOENIX": -0.15,
}

mu = np.array(list(predictions.values()))
allocations = mu / 2

for prod, pred in predictions.items():
    res.loc[prod, "Product"] = prod
    res.loc[prod, "prediction (%)"] = f"{(pred*100)}%"
    res.loc[prod, "vol (%)"] = f"{(pred / 2.0) * 100}%"
    res.loc[prod, "vol"] = (pred/2) * 1000000
    res.loc[prod, "fee"] = (pred/2)**2 * 1000000
    res.loc[prod, "pnl"] = (res.loc[prod, "vol"] * pred) - res.loc[prod, "fee"]

res["fee"] = res["fee"].astype(int)
res["vol"] = res["vol"].astype(int)
res["pnl"] = res["pnl"].astype(int)

print(f"Total Vol: {abs(res["vol"]).sum():,}")
print(f"Total Fees: {res["fee"].sum():,}")
print(f"Predicted Pnl: {res["pnl"].sum():,}")

res
```
